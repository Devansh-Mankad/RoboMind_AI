## Hopfield vs RNN

| Model | Can do | Cannot do |
|---|---|---|
| Hopfield | Associative binary-pattern recall | Temporal sequence prediction |
| RNN | Ordered sensor anomaly prediction | Hopfield-style bitmap recall |

```python
from module_d_connectionist.rnn_anomaly import train_model, predict_stream
from main import build_shift
model, accuracy = train_model()
print('Held-out accuracy:', round(accuracy, 3))
flags = predict_stream(model, build_shift(60))
print('Anomaly windows:', sum(flag for _, _, flag in flags))
```
